## Strategy selected from the rules, economy traces, and public agents

- **Land:** buy NE and SW on the proven route, normally around steps 151 and 266. Treat the $4,000 SE quadrant as conditional rather than automatic; most tested routes finish with three quadrants, and late land has too little season left to repay travel, seeds, and labor.
- **Money:** preserve capital for cheap daily hires, animal feed, and the second land unlock. Optimize the ordered market list, sell premium goods before a competing glut, reclaim shed overflow, and force end-of-season liquidation.
- **Items:** wheat remains the feed and liquidity backbone. Select animals from observed shops: sheep for yarn demand, cows for milk shops, and geese for egg shops. Carrots and tomatoes are scarcity plays only in matching shop worlds. Strawberries and melons are high-value but highly glut-sensitive, so the agent throttles and reorders those sales.
- **Algorithm:** a deterministic route/tape supplies reliable field execution; observation-driven wrappers choose the route from shop unlocks, protect seed and fertilizer budgets, estimate opponent similarity and supply, project tomato revenue, search market-order permutations, and repair overflow or late liquidation failures.

````python
import gzip, hashlib, io, tarfile
source_bytes = MAIN.read_bytes()
assert hashlib.sha256(source_bytes).hexdigest() == EXPECTED_MAIN_SHA256
ARCHIVE = OUTPUT_ROOT / 'submission.tar.gz'
buffer = io.BytesIO()
with tarfile.open(fileobj=buffer, mode='w') as tar:
    info = tarfile.TarInfo('main.py')
    info.size = len(source_bytes)
    info.mtime = 0
    info.mode = 0o644
    tar.addfile(info, io.BytesIO(source_bytes))
with ARCHIVE.open('wb') as stream:
    with gzip.GzipFile(fileobj=stream, mode='wb', mtime=0, filename='') as zipper:
        zipper.write(buffer.getvalue())
with tarfile.open(ARCHIVE) as tar:
    assert tar.getnames() == ['main.py']
    assert tar.extractfile('main.py').read() == source_bytes
print('Ready:', ARCHIVE)
print('main.py SHA256:', EXPECTED_MAIN_SHA256)
````
