# Kaggriculture: More Wheat, Smarter Sales

A farming agent combining the Metav4 production controller, order-book sales and a bounded final seven-turn collection planner. Harvest-aware fertilizer checks avoid applications that cannot improve a planned harvest. A late wheat/carrot seed budget limits purchases to remaining planting demand, with reserves for queued retries.

Version 8 adopts the independently tested seed-budget controller and retains the original market horizon of 44 turns. The final collection search uses up to 128 simulations and eight proposals per actor. Longer market horizons and extra storage/surplus rules were tested, but did not pass the full selection process. The released policy is byte-identical to Seven-Turn Rescue v2, reused here because it remained the strongest confirmed policy in this comparison.
