Write the V15 agent archive for competition submission.


```python
import base64
import hashlib
import tarfile
from pathlib import Path

ARCHIVE = Path.cwd() / "submission.tar.gz"
ARCHIVE_BYTES = base64.b64decode("")
EXPECTED_SHA256 = "5a24e730dcf5f77dcd7fb6312432ce162d5196f9fc40426b23c87effdf42ec81"
assert hashlib.sha256(ARCHIVE_BYTES).hexdigest() == EXPECTED_SHA256
ARCHIVE.write_bytes(ARCHIVE_BYTES)
with tarfile.open(ARCHIVE, "r:gz") as bundle:
    assert bundle.getnames() == ["main.py", "LICENSE.txt", "NOTICE.txt"]
print("PASS: V15 submission archive written")
print("archive SHA-256:", EXPECTED_SHA256)
print("archive bytes:", len(ARCHIVE_BYTES))
```
