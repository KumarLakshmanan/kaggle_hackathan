This is not evidence of a smooth clock penalty. It is a structural

coin margin is useful for robustness analysis, but it does not replace

## 4. Why one complete policy per seat?

Simultaneous market order resolution and engine turn order make the
two seats strategically asymmetric. A single average route can hide
that asymmetry. v26 therefore selects only at game start:

```python
actions = SEAT0_ACTIONS if obs["player"] == 0 else SEAT1_ACTIONS
```

It does **not** infer an opponent name or Notebook, and it does not
switch farms halfway through a production cycle. Each branch is a
coherent route from step 0 to step 718.

The strict-future result supports the choice:

- seat 0: **12/13**, mean margin +14,510;
- seat 1: **12/12**, mean margin +13,326;
- Top-5 cases: **4/4**;
- all v23-opening cases: **9/9**.

This is a portfolio over public strategic memory, not an identity
lookup table. Exact byte-for-byte copies remain symmetric in
expectation across randomized seats; no deterministic public agent
can manufacture a lasting secret edge against its own exact copy.

## 5. The closed loop is deliberately sparse

The route screen shows where the gain came from. On the same inner
panel, adding the market overlay changed:

- Yubo: 40/46 → **41/46**;
- Gbining: 39/46 → **40/46**.

Most alpha is route refresh, not a large universal planner. The
retained feedback has only two event types:

1. if a planned BUILD/PLANT actor actually stands on a WEED, DIG and
   replay that actor's bounded local trace;
2. if a turn already contains at least two SELLs, rank only those
   SELL slots by official nonlinear price impact and current Town
   demand recovery.

Ordinary turns create, delete, or resize no SELL. BUY, HIRE, land,
animal, seed, farmer, and hand plans stay in their complete route.
The residual-capital market-maker expert is not included: on the
preceding fresh panel it lost route coherence and did not clear the
win gate. Complexity is retained only when it changes wins after time
separation.

[Structured Economic Policy](https://www.kaggle.com/code/pilkwang/kaggriculture-structured-economic-policy)
is useful context for reserve-aware economic control; this release
takes the narrower experimentally supported subset.

## 7. The loss and the missing test are visible

The only strict-future loss was:


More important than hiding 24/25 behind a percentage: the later

does replay evaluation establish robustness to an adaptive opponent.

Two tempting shortcuts were rejected:

- freezing an adaptive Top-1/Top-2 replay as if it were a portable
  policy; a trace records one realized state path, not its branch logic;
- same-turn queue permutation selected on a tiny tape/seed set.
