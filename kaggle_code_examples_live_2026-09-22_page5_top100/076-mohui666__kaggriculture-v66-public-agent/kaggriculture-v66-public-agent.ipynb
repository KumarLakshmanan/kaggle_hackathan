## What v66 is — and is not

v66 is a lightweight, deterministic rule agent. It reads only the observation exposed to the current player, selects among several economic routes, and sanitizes the final structured action before returning it.

It is **not** a PPO/GAE model, does not ship neural weights, and does not require PyTorch, CUDA, a GPU, network access, or an inference service. Runtime dependencies are Python standard-library modules only.

## Decision flow

```text
public observation
    |
    +-- base production / movement / purchase / sale routes
    +-- public-state signature
    |      +-- shop reveal order
    |      +-- public money and assets
    |      +-- market inventory and prices
    +-- stage-aware route selection
    |      +-- recovery
    |      +-- known-yarn
    |      +-- bakery / pizza
    |      +-- curve / clone response
    +-- endgame return, harvest, and realization
    +-- sanitize_action -> Kaggle action
```

The strategy is not a general-purpose search engine. It combines a mature closed-loop baseline with narrowly triggered public-state branches; unmatched states continue through the tested baseline path.
