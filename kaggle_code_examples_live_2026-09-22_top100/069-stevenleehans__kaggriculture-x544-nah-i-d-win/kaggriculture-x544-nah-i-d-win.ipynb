## Full-season official-engine smoke

This checks both seats for 720 turns.

```python
from kaggle_environments import make

smoke_results = []
for x567_seat in (0, 1):
    agents = [str(main_path), "starter"] if x567_seat == 0 else ["starter", str(main_path)]
    env = make(
        "kaggriculture",
        configuration={"episodeSteps": 720, "seed": 567567},
        debug=False,
    )
    env.run(agents)
    final = [(state.status, state.reward) for state in env.steps[-1]]
    assert all(status == "DONE" for status, _ in final), final
    smoke_results.append({"x567_seat": x567_seat, "final": final})

print("Official Kaggriculture smoke: PASS")
print(smoke_results)
```
