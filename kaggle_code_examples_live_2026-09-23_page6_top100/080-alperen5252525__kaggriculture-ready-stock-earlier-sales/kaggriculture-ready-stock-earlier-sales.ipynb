# Kaggriculture: Ready Stock, Earlier Sales

A runnable public-source agent with one focused experiment: keep the production
program and funded opening, but give already available cash products more chances
to sell before the shared market fills.

Our selected configuration **advance6** finished **56 wins / 8 losses / 0 ties**
in a separate 64-game local panel. The previous release finished
**48 / 16 / 0**; unchanged V46 finished
**42 / 8 / 14**.

These are local match outcomes, not a prediction of 2600, 2700 or any live rating.
Run the build cell below to obtain a single-file submission; no GPU or external
dataset is needed and nothing is submitted automatically.

## What changes—and what does not

The candidate is built on Ahmed Berat Özer's public V46. Its production, worker,
warehouse and funded opening programs are retained. We screened unchanged V46,
6- and 12-turn sale lookahead, immediate clone reservation and removal of the
delayed wheat lift, alongside our previous release. We selected by match points,
then worst-opponent match points, then cash margin before opening the holdout.

The selected configuration is `advance6`. Six-turn lookahead changes only
`_ADV_LOOK` from 3 to 6: the existing runtime guards still check physical stock,
pickup obligations, market slots, dawn boundaries and protected funding sales.
This is a parameter experiment on credited public mechanisms—not a new farming
engine or an undisclosed private-leader algorithm.

## Separate confirmation panel

Eight previously unused initial seeds × four public policies × both seats.

| Opponent | Wins / losses / ties | Mean cash margin |
|---|---:|---:|
| v46 | 15 / 1 / 0 | +347.8 |
| reactive7 | 13 / 3 / 0 | +239.1 |
| jaxa | 13 / 3 / 0 | +239.1 |
| Pipe8 | 15 / 1 / 0 | +839.2 |

The opponents share substantial ancestry. In our screen Reactive V7 and Jaxa
produced identical outcomes, so do not interpret four notebook names as four
independent elite strategy families. Initial seeds also do not guarantee identical
realized shop draws after policies diverge. The panel can find a current-public-
family edge; it cannot establish durable ladder superiority.

Two full official-engine smoke checks completed all 719 calls per player, both
agents DONE, with exact rewards when their joint action streams were replayed
in the native simulator. Raw summaries and smoke-check evidence are included below.
