## One opening to rule the board

The first check is identity: the turn-24 opening signature. Every one of the six corpora plays a **single** opening at 100 % of its episodes, and it is **the same opening in all six**.

## The chassis is multi-route, and the tuning knob is the route mix

The public build carries more than one full-season route behind that shared opening and selects one early, by game state. Two portfolios dominate everywhere: **route A, melon 20 / strawberry 34**, and **route B, melon 12 / strawberry 42** (a 12/34 hybrid appears in small numbers in Tschinkel's corpora).

Every corpus MIXES the two. What separates the submissions is the **distribution of that choice**, and it is the one thing that visibly moved across the balance change:

* カワシギ pre-update: 69 % route A. カワシギ's current king (3,196): **89 % route B**.
* Tschinkel was already B-leaning before the update (64 %) and stays B-leaning after (72 %).
* カワシギ's newest bot mixes almost 50/50, like the public build does out of the box.

So the story is not one team copying the other's constants: both run the shared chassis, and the winning tweak, first by Tschinkel and then harder by カワシギ, is **biasing the route choice toward B**. Route B is the absorption-shaped portfolio: less melon, whose book has little room, more strawberry, whose book is deep.

```python
order = [(g, seat) for g in G for seat in ("0", "1")
         if not (g["seats"].get(seat) or {}).get("thin")
         and "portfolio_combos" in (g["seats"].get(seat) or {})]
labels, aa, bb, other = [], [], [], []
for g, seat in order:
    c = g["seats"][seat]["portfolio_combos"]
    n = sum(c.values())
    labels.append(f"{g['label']} s{seat} (n={n})")
    aa.append(c.get("20/34", 0) / n)
    bb.append(c.get("12/42", 0) / n)
    other.append(1 - aa[-1] - bb[-1])
y = np.arange(len(labels))
fig, ax = plt.subplots(figsize=(10, 0.34 * len(labels) + 1.2))
ax.barh(y, aa, color="#B45309", label="route A: melon 20 / strawberry 34")
ax.barh(y, bb, left=aa, color="#0F172A", label="route B: melon 12 / strawberry 42")
ax.barh(y, other, left=np.array(aa) + np.array(bb), color="#DCEEE1",
        label="hybrids")
ax.set_yticks(y); ax.set_yticklabels(labels, fontsize=7)
ax.invert_yaxis(); ax.set_xlim(0, 1)
ax.set_xlabel("share of episodes")
ax.legend(fontsize=7, loc="lower right")
ax.set_title("route choice by corpus and seat", fontsize=9)
plt.tight_layout(); plt.show()
```

Each panel is one corpus and seat; each **row is one real episode**. ## Is the dark mass a second base plan?

The X-rays invite a specific hypothesis: maybe the divergent episodes are not adapting at all, maybe they are playing a SECOND fixed plan, different from the mode but identical to each other. If true, the right move for anyone studying these agents would be to extract that hidden plan. The hypothesis is testable with two regroupings, and it fails in its strong form.

First test: recompute each corpus's divergence WITHIN its portfolio groups instead of against the pooled mode. If the dark were mostly route branching, divergence should collapse once episodes are compared only against their own route's mode. It barely moves: the drop is a few points at most, everywhere.

Second test: at every divergent cell, classify the minority. An episode whose alternative action is shared by other episodes at that turn is evidence of a discrete branch; an alternative played by that episode alone is per-game adaptation. **Roughly two thirds of the minority mass is singletons**, unique to one episode at that turn.

So the dark is not a hidden plan. What discrete structure exists, the shared third, matches the route variants that are already visible in the published builds themselves: the same chassis carries several full-season routes selected by the town's shop draw, and those routes are public. The base plans are common knowledge; **the edge at the top is the per-game, intraday adaptive layer**, which is exactly the part a plan-extraction approach cannot capture.

```python
print(f"{'corpus':<26}{'seat':>5}{'pooled':>8}{'within groups':>32}"
      f"{'minority shared':>17}{'singleton':>11}")
for g in G:
    for seat in ("0", "1"):
        s = g["seats"].get(seat) or {}
        b = s.get("branching")
        if not b:
            continue
        w = "  ".join(f"{k}:{v['divergence']:.3f}(n={v['n']})"
                      for k, v in sorted(b["within_portfolio"].items()))
        sh, si = b["minority_shared"], b["minority_singleton"]
        tot = max(1, sh + si)
        print(f"{g['label']:<26}{seat:>5}{b['pooled_divergence']:>8.3f}"
              f"{w:>32}{100*sh/tot:>16.0f}%{100*si/tot:>10.0f}%")

fig, ax = plt.subplots(figsize=(10, 3.4))
labels, sing = [], []
for g in G:
    for seat in ("0", "1"):
        b = (g["seats"].get(seat) or {}).get("branching")
        if b:
            labels.append(f"{g['label']} s{seat}")
            tot = max(1, b["minority_shared"] + b["minority_singleton"])
            sing.append(b["minority_singleton"] / tot)
y = np.arange(len(labels))
ax.barh(y, sing, color="#0F172A", label="singleton: unique to one episode")
ax.barh(y, [1 - x for x in sing], left=sing, color="#B45309",
        label="shared alternative: a discrete branch")
ax.set_yticks(y); ax.set_yticklabels(labels, fontsize=7)
ax.invert_yaxis(); ax.set_xlim(0, 1)
ax.set_xlabel("composition of the divergent minority")
ax.legend(fontsize=7, loc="lower right")
plt.tight_layout(); plt.show()
```

## The green stripes, because the question asks itself

Looking at the X-rays, regularly spaced green bands run through even the darkest late-game regions: the divergence does not propagate to the bell as one widening front, it comes in blocks. The period is the in-game day, and the two cells below measure what the stripes are.

First, folding every divergence matrix by hour of day shows the anchor is **hour 1**, not hour 0. Hour 0 is the decision hour: the daily sells, the feed purchases, the hire block, all of which differ between episodes. By hour 1 the engine has reset every unit to its spawn, dissolved yesterday's hands and rehired, and every episode executes the day-script's first move from the same configuration.

Second, divergence is sticky within a day and only partly survives the night: the probability that a divergent episode stays divergent is about 0.95 from one turn to the next in mid-day, and drops to about 0.61 across the day boundary. The night is a damper, not a barrier, because farm state (crops, herd, shed, money) carries over even though positions do not.

So the day is the module. These agents are not one 720-step trajectory but thirty day-programs chained through a narrow synchronisation bottleneck, and their adaptation lives inside the day.

```python
hours = (np.arange(1, 719) % 24)
fold = np.zeros(24); cnt = np.zeros(24)
night_surv = []; day_surv_num = day_surv_den = 0
for g in G:
    for seat in ("0", "1"):
        s = g["seats"].get(seat) or {}
        if s.get("thin") or "plan_rows_hex" not in s:
            continue
        div = (unpack(s["plan_rows_hex"], s["n"]) |
               unpack(s["market_rows_hex"], s["n"])).astype(bool)
        for h in range(24):
            sel = div[:, hours == h]
            fold[h] += sel.sum(); cnt[h] += sel.size
        for day in range(1, 29):
            prev = div[:, day*24 - 3:day*24 - 1].any(axis=1)
            nxt = div[:, day*24:day*24 + 2].any(axis=1)
            if prev.sum() >= 3:
                night_surv.append((nxt[prev].mean(), int(prev.sum())))
        mid, prv = div[:, 1:], div[:, :-1]
        hsel = (np.arange(2, 719) % 24) > 3
        day_surv_num += (mid & prv)[:, hsel[:mid.shape[1]]].sum()
        day_surv_den += prv[:, hsel[:prv.shape[1]]].sum()

rate = fold / np.maximum(1, cnt)
fig, ax = plt.subplots(figsize=(10, 3))
colors = ["#0F172A" if h != 1 else "#2E7D32" for h in range(24)]
ax.bar(range(24), rate, color=colors)
ax.set_xticks(range(24)); ax.set_xlabel("hour of the in-game day")
ax.set_ylabel("share of episode-turns diverging")
ax.set_title("divergence folded by hour: hour 1 is the resynchronisation anchor",
             fontsize=9)
plt.tight_layout(); plt.show()

w = np.array([n for _, n in night_surv]); v = np.array([x for x, _ in night_surv])
print(f"divergence persistence, turn to turn in mid-day: "
      f"{day_surv_num/max(1,day_surv_den):.2f}")
print(f"divergence persistence across the night:          "
      f"{(v*w).sum()/w.sum():.2f}")
```

## What I take from it

1. **The top of this ladder is one family.** Six corpora, both teams, both eras, one opening, and it is the public consensus backbone. Above 3,000 this is a contest of mutation quality on a shared public chassis.
2. **The tuning knob that moved is the route mix.** The chassis carries two portfolios; the king plays the absorption-shaped one 89 % of the time, Tschinkel leaned that way first, and the out-of-the-box 50/50 mix is what the weakest of the four current bots plays.
3. **There is no hidden second plan.** Regrouping by portfolio explains only a few points of divergence and about two thirds of the divergent minority is unique to a single episode. The discrete part matches the route variants already published in the builds; the rest is per-game adaptation.
4. **The adaptation is intraday and the day is the module.** Hour 1 is the resynchronisation anchor, divergence persists at ~0.95 turn to turn inside a day and only ~0.61 across the night. Variation does not propagate to the bell as one front; it lives in daily compartments.
5. **Two adaptation philosophies coexist at the same rating**: a per-game feed loop against a nearly scripted plan with composition tuning. The newest Tschinkel bot is 24-0 but hours old.
6. **Nobody at the top entered the convex goods.** Whatever the balance change was worth, the measured answer of the best agents so far is portfolio resizing inside the old goods.
