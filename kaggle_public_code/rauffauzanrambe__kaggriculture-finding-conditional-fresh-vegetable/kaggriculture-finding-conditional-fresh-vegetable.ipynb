### Why Replay Analysis Demands Caution

A replay captures a single match, not the opponent’s underlying strategy. A losing player might execute a superior policy, only to be compromised by an unfavorable random seed or execution order in the market. Consequently, I treat moves from both players as empirical evidence rather than strictly classifying losing actions as blunders.

### Objective Function vs. Evaluation Metric

Kaggriculture is a 720-turn, two-player zero-sum simulation. Leaderboard rankings are determined strictly by **win/loss/tie skill ratings**, meaning a massive coin margin grants no additional rating points over a narrow win. Local bank totals and TrueSkill ratings answer fundamentally different questions; optimizing purely for endgame wealth does not guarantee competitive resilience.

### Action economy beats the crop table

Each unit (farmer + hired hands) gets **one field op per turn**.  
A high-margin crop that costs two extra walks often loses to a weaker crop next to the shed.

**Rule of thumb:** hiring about 10 hands costs roughly **$143 for a full day** of parallelism. Under-hiring early is usually more expensive than over-hiring inside the cheap Fibonacci region.


````python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


def fibonacci(n: int) -> int:
    """Calculates the n-th Fibonacci number iteratively.

    Args:
        n (int): Index of the Fibonacci number.

    Returns:
        int: Fibonacci value at the n-th index.
    """
    a, b = 1, 1
    for _ in range(n):
        a, b = b, a + b
    return a


def calculate_hire_day_cost(n_hands: int) -> int:
    """Calculates the total cost to hire n workers (hands) today.

    Args:
        n_hands (int): Number of workers to hire.

    Returns:
        int: Total hiring cost in coins.
    """
    return sum(fibonacci(i) for i in range(n_hands))


# -----------------------------------------------------------------------------
# 1. Calculation of Hiring Cost Data
# -----------------------------------------------------------------------------
hands_range = np.arange(1, 17)
hire_costs = [calculate_hire_day_cost(int(n)) for n in hands_range]

# -----------------------------------------------------------------------------
# 2. Visualization of Hiring Cost Curve (Matplotlib)
# -----------------------------------------------------------------------------
fig, ax = plt.subplots(figsize=(9, 4.5))

ax.plot(
    hands_range,
    hire_costs,
    marker="o",
    linestyle="-",
    linewidth=2,
    color="#0f766e",
)

# Cost region threshold lines
ax.axvline(
    x=10,
    color="#f59e0b",
    linestyle="--",
    label="~10 hands (End of cheap region)",
)
ax.axvline(
    x=12,
    color="#dc2626",
    linestyle="--",
    label="~12 hands (Steep cost surge)",
)

# Label and title settings
ax.set_xlabel("Hands Hired Today")
ax.set_ylabel("Total Hire Cost (Coins)")
ax.set_title("Fibonacci Hire Curve (Default Mult = 1)")
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# -----------------------------------------------------------------------------
# 3. Data Summary Output (Pandas DataFrame)
# -----------------------------------------------------------------------------
df_cost_summary = pd.DataFrame(
    {"hands": hands_range, "day_cost": hire_costs}
)

# Filter sample data at key points
selected_hands = [6, 8, 10, 11, 12, 14]
filtered_df = df_cost_summary.query("hands in @selected_hands")

print(filtered_df.to_string(index=False))
````

## 2. Engine details that separate weak agents from strong ones

### 2.1 Melon yield window (trust the engine, not only the overview table)

For one-time crops, watering during a bonus window adds yield immediately:

* **Yield Window Calculation**: Melon `max_yield_day = 12` implies a window start of $\lceil 12/2 \rceil = 6$, covering ages **6 to 12**.
* **Base Yield**: A melon starts at `yield_units = 1`.
* **Watering Bonus**: Each watered day within the window adds **+1** yield (or **+2** if fertilized).
* **Yield Cap**: Yield is capped at **6**, meaning continuous watering reaches maximum yield around **age 10**, well before age 12.
* **Potential Output**: $16\text{ tiles} \times 6 = 96\text{ melons}$ under optimal execution.
* **Common Pitfall**: Missing water during days 6–10 causes a silent yield leak (e.g., yielding ~70 units instead of 96).

### 2.2 Inventory vs. Shed dynamics (`SELL` vs. `DROP`)

* `HARVEST` places items directly into the **unit inventory**, not the shed.
* `SELL` executes transactions exclusively using items stored in the **shed**.
* **Action Blocking**: If you harvest 90 melons but fail to perform a `DROP` while adjacent to the shed, the market will not register them until the end-of-day auto-drop. This delay is often **too late** to fund same-turn land or animal purchases.
* **Shed Capacity Lock**: Keep in mind that a full shed will cause `BUY_PRODUCT` and `BUY_SEED` actions to fail silently, which can inadvertently starve animals if feed cannot be delivered.

### 2.3 Fertilizer as a liquid cash asset

* While competition documentation emphasizes buying fertilizer for crop boosts, the engine's generic `SELL` handler accepts fertilizer sales.
* **Passive Income**: Animals generate fertilizer daily, creating a reliable secondary cash flow stream prior to market saturation.

### 2.4 Market saturation and price decay

* Selling large quantities of crops or animal products in a short window triggers market price decay.
* **Staggered Liquidation**: Strong agents monitor market price curves and batch their `DROP` + `SELL` routines to avoid selling high-volume harvests into a depressed market.

### 2.5 Action economy and turn efficiency

* Move actions and inventory management actions (`DROP`, `PICKUP`) consume valuable turn steps within the 24-turn daily cycle.
* Pathfinding should minimize unnecessary trips to the shed by optimizing harvest routes and combining drop-offs with seed/fertilizer pick-ups.

### 2.6 Ladder scoring and evaluation metrics

* **True Objective**: Match ratings update strictly from **win / loss / tie** outcomes.
* **Bank balance misdirection**: A 140k bank balance that loses the game still results in a rating penalty.
* **Evaluation Strategy**: While local mean bank balance against `starter` bots is a useful filtering benchmark, optimizing purely for cash accumulation does not guarantee victory against competitive agents.

````python
import matplotlib.pyplot as plt
import numpy as np

# --- Market Parameters ---
I0 = 10000          # Initial inventory
BASE = 250          # Base price
T = 300             # Scale parameter
ABOVE_TARGET = 3.6  # Multiplier above target
SEED_COST_PER_TILE = 80
UNITS_PER_TILE = 6


def melon_price(inv: float) -> int:
    """Calculates the melon price based on the current inventory level."""
    x = max(0.0, float(inv) - I0)
    amp = ABOVE_TARGET * BASE / (T**2)
    return max(1, int(round(BASE - amp * x * x)))


def calculate_glut_curve():
    tiles = np.arange(1, 31)
    marginal = []
    inv = I0

    for _ in tiles:
        rev = 0
        for _ in range(UNITS_PER_TILE):
            p = melon_price(inv)
            rev += p
            if p > 1:
                inv += 1
        marginal.append(rev - SEED_COST_PER_TILE)

    cum = np.cumsum(marginal)
    best_idx = int(np.argmax(cum))
    best_tile = int(tiles[best_idx])

    return tiles, marginal, cum, best_tile


# --- Perform Calculation ---
tiles, marginal, cum, best = calculate_glut_curve()

# --- Data Visualization ---
fig, axes = plt.subplots(1, 2, figsize=(12, 5), dpi=100)

# Plot 1: Marginal Net Value
bar_colors = ["#0f766e" if v > 0 else "#e2e8f0" for v in marginal]
axes[0].bar(tiles, marginal, color=bar_colors, edgecolor="#0f766e", linewidth=0.5)
axes[0].axhline(0, color="#334155", lw=1, linestyle="--")
axes[0].set_title("Marginal Net Value of Next Melon Tile", fontsize=12, fontweight="bold")
axes[0].set_xlabel("Tile # Sold into One Glut Path", fontsize=10)
axes[0].set_ylabel("Revenue - Seed Cost (Coins)", fontsize=10)

# Plot 2: Cumulative Net Value
axes[1].plot(tiles, cum, lw=2.5, color="#0f766e", label="Cumulative Net")
axes[1].scatter(
    [best],
    [cum[best - 1]],
    s=100,
    color="#dc2626",
    zorder=5,
    label=f"Peak Optimal (~{best} Tiles)",
)
axes[1].set_title("Cumulative Net Value (Labor Ignored)", fontsize=12, fontweight="bold")
axes[1].set_xlabel("Melon Tiles", fontsize=10)
axes[1].set_ylabel("Cumulative Net Coins", fontsize=10)
axes[1].legend(frameon=True, facecolor="white", framealpha=0.9)

# Additional Styling for Both Axes
for ax in axes:
    ax.grid(True, linestyle=":", alpha=0.6)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()

# --- Summary Output ---
print({
    "illustrative_peak_tiles": best,
    "max_cumulative_profit": cum[best - 1],
    "note": "Labor and rival sales move the real optimum; public leaders often use ~6-16 melons, not a full 25-tile dump.",
})
````

## 3. Strategy Cluster Descriptions

### **Cluster A: Pure Cow Ranch**

* **Core Philosophy:** Aggressive early-game focus on livestock over crops to force opponents into an immediate efficiency race.
* **Mechanism:** Opens directly with 3–10 cows and minimal to no crops, quickly building up to a peak herd of roughly 10 cows (frequently utilizing North/East placement strategies).
* **Strengths:** Heavily punishes slow-building crop bots or "soft" strategies that lack early defense or capital depth.
* **Failure Mode:** Extreme susceptibility to "milk wars." If multiple bots run this strategy simultaneously, oversupply causes milk prices to crash, leading to bank collapse and coin-flip Head-to-Head (H2H) outcomes.

### **Cluster B: Melon IPO**

* **Core Philosophy:** High-risk, high-reward capital injection strategy centered around quick crop liquidation.
* **Mechanism:** Plant a massive 16-melon crop alongside 2 initial cows. Around Day 10, execute a complete farm dump to transition into 12 cows, 8 sheep, and 3 quads.
* **Strengths:** Generates a massive early capital spike, allowing the bot to snowball infrastructure faster than any other cluster when left uncontested.
* **Failure Mode:** Heavily reliant on market exclusivity. If a second bot attempts a melon dump at the same time, crop prices hit the $1 price floor, destroying expected returns.

### **Cluster C: Staged Economic Herd (C0x Family)**

* **Core Philosophy:** Methodical, multi-stage scaling designed to balance cash flow with herd expansion.
* **Mechanism:** Establishes a strict baseline floor of **3 cows + 1 sheep**, then systematically scales through structured phases: 4 to 8 cows, 10 to 14 cows, and eventually 15 cows.
* **Strengths:** Currently forms the backbone of the strongest public code bases due to its predictable, optimized expansion timing.
* **Failure Mode:** Systemic risk through popularity. Because it is widely copied and forked, multiple C0x bots in the same lobby trigger correlated market losses.

### **Cluster D: Adaptive Leader Style**

* **Core Philosophy:** Dynamic resource reallocation based on real-time market signals rather than fixed build orders.
* **Mechanism:** Starts with 3 cows, 1 sheep, and approximately 6 melons, dynamically shifting the mix between cows, sheep, and geese based on current commodity prices.
* **Strengths:** Highly resilient to market crashes; seamlessly rebalances away from milk whenever cow-heavy strategies saturate the market.
* **Failure Mode:** High code complexity makes it significantly harder to clone, fine-tune, or execute without logic bugs.

## 4. Replay Diary: From c14 to the Current Multi-Leader Market Meta

### **4.1 The First Useful Replay Rule**

* **Core Philosophy:** Moving away from chasing single high-scoring "spectacular" replays toward identifying repeatable, deterministic execution patterns that hold up across different match conditions.
* **The Key Shift:** Overfitting to a single winner's replay introduces noise. True optimization requires filtering for structural consistency across multiple opponents before attempting to distill or copy a strategy tape.

#### **The Three-Point Validation Framework:**

Before distilling any replay into a production agent, the run must pass three strict criteria:

1. **Field Schedule Consistency:** Does the bot execute the exact same tile placement, crop rotation, and livestock expansion sequence against different opponents?
2. **Market Adaptability:** Is the market transaction schedule stable, or is it overly sensitive to opponent-induced price fluctuations?
3. **Paired-Seat Equivalence:** Does the distilled strategy win regardless of starting seat position (Seat 0 vs. Seat 1)?

### **Case Study: The Evolution of `c14**`

* **Origin (Senkin13 Refresh):** The `senkin13` baseline passed all validation tests unusually cleanly. Across 5 separate matches, it displayed an identical 720-turn field and market execution path.
* **Final Target State:** The run settled into an optimal equilibrium of **~8 cows, 6 sheep, and 7 strawberry plots**.
* **The Refinement (`c14` Upgrade):** Added a specialized, observation-driven cleanup routine active only during the **final 8 turns** to liquidate remaining assets and maximize end-game liquid cash.

#### **Validation Results & Performance Impact:**

* **Head-to-Head Gate Test:** `c14` dominated previous iterations (`c11`, `c12`, and `c13`) with a decisive **27–3 win ratio** in paired-seat matches.
* **Ladder Standings:** Achieved a peak rating of **2182.2** on its standalone ladder run.

> **Key Takeaway:** Prioritizing replay stability and repeatable tape distillation yields significantly higher competitive win rates than optimizing for isolated peak scores against weak bots.

### **4.2 Hamburger: Changing the Question from *What to Build* to *When to Sell***

* **Core Philosophy:** Shifting the strategic focus from layout and build optimization (*what to build*) to precise market execution timing (*when to sell*) to gain a decisive edge in mirror matches.
* **Mechanism (`Clone Quad H1`):** * Continuously checks whether both competing bots are following nearly identical farm trajectories.
* After validating trajectory alignment across two consecutive checkpoints, it looks **one turn ahead** in its internal schedule.
* Executes a preemptive sale of a single high-value commodity line (**melon, strawberry, milk, or wool**) exactly **one turn before** the anticipated shared-market dump.


* **Strategic Impact:** * Captures peak market prices immediately prior to liquidity drops and price floor collapses caused by simultaneous clone dumps.

### **Integration & Progression (`c15`)**

* **Notebook Results:** The original Hamburger wrapper achieved a dominant **6–0 victory record** against its anchor bot, maintaining a mean score margin of **+1,865.7 points**.
* **The Hybrid Upgrade (`c15`):** * Ported Hamburger’s **one-turn front-running logic** directly onto the structurally superior **refreshed Senkin baseline** (from `c14`).
* The resulting hybrid, **`c15`**, paired elite production scaling with micro-timed market front-running.



> **Key Takeaway:** When two competing bots run virtually identical farm layouts, game-winning point differentials are no longer decided by infrastructure, but by out-timing opponent market dumps by a single turn.

## 8. Local evaluation protocol

```text
1. Compile/import the exact packaged main.py
2. Self-play validation: both agents must finish DONE
3. Play the incumbent and the unmodified parent on disjoint seeds, both seats
4. Treat a loss to either as a veto, even if aggregate wins look strong
5. Play several genuinely different public field families
6. Report per-opponent records, not only the aggregate
7. Freeze all parameters, then run one untouched final seed block
8. Audit cash, feed, animal survival, shed overflow and runtime
9. Keep the result files, including losses
10. Only then build an archive
```

Shared-market games are not symmetric. A one-seat test can reverse the
apparent winner.


````python
def h2h(agent_a, agent_b, seeds=range(3), both_seats=True):
    """Quick head-to-head helper. agent_* may be callables or built-in names."""
    seats = (0, 1) if both_seats else (0,)
    rows = []
    for seed in seeds:
        for seat in seats:
            env = make("kaggriculture", configuration={"episodeSteps": 720, "seed": int(seed)})
            order = [agent_a, agent_b] if seat == 0 else [agent_b, agent_a]
            env.run(order)
            final = env.steps[-1]
            r0, r1 = final[0].reward, final[1].reward
            me, opp = (r0, r1) if seat == 0 else (r1, r0)
            rows.append({
                "seed": seed,
                "seat": seat,
                "me": me,
                "opp": opp,
                "win": me > opp,
                "status0": str(final[0].status),
                "status1": str(final[1].status),
            })
    df = pd.DataFrame(rows)
    summary = {
        "games": len(df),
        "wins": int(df.win.sum()),
        "win_rate": float(df.win.mean()),
        "mean_bank": float(df.me.mean()),
        "mean_opp": float(df.opp.mean()),
    }
    return df, summary


# Example using built-ins only (swap in your agent callables locally)
df, summary = h2h("starter", "random", seeds=range(2))
print(summary)
df
````
