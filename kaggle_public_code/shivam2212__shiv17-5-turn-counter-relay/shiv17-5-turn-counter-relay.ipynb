<div id="game" style="background:#ffffff;border:1px solid #e5e7eb;border-left:3px solid #0f766e;border-radius:6px;padding:20px 24px;margin:0 0 16px 0;font-family:Arial,Helvetica,sans-serif;">
  <h2 style="margin:0 0 10px 0;color:#111827;font-size:19px;font-weight:700;">1 &middot; The game in 30 seconds</h2>
  <p style="margin:0 0 10px 0;font-size:14px;line-height:1.7;color:#374151;">
    Kaggriculture is a 1v1, turn-based farming economy. Each player runs a 10&times;10 farm
    over a fixed 720-turn season (30 days &times; 24 hourly steps), moving farmers and hired
    hands, managing a 100-slot shed, and batching market orders.
  </p>
  <ul style="margin:0;padding-left:20px;font-size:14px;line-height:1.8;color:#374151;">
    <li><b>Shared dynamic market.</b> Prices move with aggregate inventory through per-resource shape functions (linear, sq, sqrt, log), with a hard $1 floor.</li>
    <li><b>Win condition.</b> Most coins at season end.</li>
    <li><b>Asymmetric information.</b> Both farms are visible every turn; sheds, seeds and inventories are private.</li>
  </ul>
  <p style="margin:10px 0 0 0;font-size:13px;line-height:1.6;color:#6b7280;">
    Because supply is shared and visible, <i>when</i> you sell matters as much as <i>what</i>
    you sell &mdash; dumping into a glut returns $1/unit, selling into a thin market can return $100+.
  </p>
</div>

  <h2 style="margin:0 0 10px 0;color:#111827;font-size:19px;font-weight:700;">3 &middot; The pattern &mdash; boatlee V16-RC2's fixed 3-turn relay</h2>
  <p style="margin:0 0 10px 0;font-size:14px;line-height:1.7;color:#374151;">
    Reverse-engineering <code style="background:#f3f4f6;border:1px solid #e5e7eb;border-radius:4px;padding:1px 5px;">boatlee_v16rc2/main.py</code>
    surfaced a guarded, debt-accounted fertilizer market relay:
  </p>
  <ul style="margin:0 0 12px 0;padding-left:20px;font-size:14px;line-height:1.8;color:#374151;">
    <li>At steps 216 / 240 / 264 it compares a public farm-signature distance between the two farms. If the opponent is a near-mirror (distance &le; 8), it locks on.</li>
    <li>For every scheduled fertilizer sale that would land at <code style="background:#f3f4f6;border:1px solid #e5e7eb;border-radius:4px;padding:1px 5px;">future_step % 24 == 17</code>, it sells all its fertilizer 3 turns early &mdash; at <code style="background:#f3f4f6;border:1px solid #e5e7eb;border-radius:4px;padding:1px 5px;">step % 24 == 14</code>.</li>
    <li>Effect: boatlee books a decent price for itself, then the aggregate inventory spike crashes the shared price to $1 right before the mirror's scheduled sell turn.</li>
  </ul>

  <h2 style="margin:0 0 10px 0;color:#111827;font-size:19px;font-weight:700;">4 &middot; The counter &mdash; a 5-turn lead</h2>
  <p style="margin:0 0 10px 0;font-size:14px;line-height:1.7;color:#374151;">

    it with a longer lead (4 or 5 turns), it sells first, captures the peak price, and

  </p>

    fixed lead with a 5-turn counter-relay (<code style="background:#f3f4f6;border:1px solid #e5e7eb;border-radius:4px;padding:1px 5px;">_RC3_RELAY_LEAD = 5</code>, triggering at

  </p>

  <p style="margin:0;font-size:13.5px;line-height:1.6;color:#6b7280;">Conceptual only &mdash; not the shipped agent's guard logic, repayment bookkeeping, or route.</p>

```python
# The whole winning idea, in 10 lines (conceptual -- not the shipped agent)
RELAY_LEAD = 5                      # sell 5 turns before the scheduled sale, not 3

def should_counter_relay(step):
    # True when our dump lands 2 turns before boatlee's fixed relay
    future_step = step + RELAY_LEAD          # where our sale would land
    return (future_step % 24) == 17          # boatlee's scheduled-sell slot

# Which turns fire? (step 12 with lead 5 -> lands at 17; boatlee dumps at 14)
for step in range(10, 18):
    print(f"step {step:2d} (mod24=={step % 24:2d}): dump={should_counter_relay(step)}")
```

  <h2 style="margin:0 0 10px 0;color:#111827;font-size:19px;font-weight:700;">6 &middot; Verification &mdash; gate family + fresh seeds</h2>
  <p style="margin:0 0 10px 0;font-size:14px;line-height:1.7;color:#374151;">
    Our first pass at this notebook benchmarked on a frozen 8-seed gate family (16 games
    per opponent) and reported <b>93.8% (15W/1L)</b> against both boatlee and kaitofukami.
    Fourteen of those sixteen margins came out to an identical <code style="background:#f3f4f6;border:1px solid #e5e7eb;border-radius:4px;padding:1px 5px;">+176</code>
    (or <code style="background:#f3f4f6;border:1px solid #e5e7eb;border-radius:4px;padding:1px 5px;">+213</code> vs kaitofukami) &mdash; which reasonably looked like a
    copy-pasted placeholder rather than real output.
  </p>
  <p style="margin:0 0 10px 0;font-size:14px;line-height:1.7;color:#374151;">
    It isn't a placeholder. The repeated value is a real engine artifact: on near-mirror
    match-ups with fixed 719-step tapes, seed-invariant price capture makes several seeds
    converge on the exact same margin, and mirror symmetry makes seat 0 and seat 1 identical
    on those seeds. We reproduced the gate-family logs exactly on rerun, found matching
    constant margins in pre-existing repo logs that predate this notebook, and then ran
    <b>16 additional, never-before-used seeds</b> to check whether the pattern held outside
    the original gate family.
  </p>
  <p style="margin:0 0 10px 0;font-size:14px;line-height:1.7;color:#374151;">
    The fresh seeds confirmed the mechanism is real (constants reappear at seeds 3434343,
    5656565, 6666666 and 9999999 with absolute scores ranging 80K&ndash;138K) but also surfaced
    something the gate family hid: <b>seat-dependence on one seed.</b> Seed 1212121 wins big
    as seat 0 (+7,379 / +7,411) but loses as seat 1 (&minus;7,063 / &minus;7,034) against both
    opponents. Net per seed is still positive, so ShiV17 remains undefeated at
    <i>seed level</i>, but the clean 93.8% <i>game-level</i> number doesn't survive contact with
    a broader seed set.
  </p>
  <div style="background:#fffbeb;border:1px solid #fde68a;border-radius:6px;padding:10px 14px;color:#92400e;font-size:13px;line-height:1.6;">

```python
# ---- Chart 1: game-level win rate, gate-only vs combined (gate+fresh) ----
labels = ["vs boatlee\n(gate only, 16g)", "vs boatlee\n(combined, 32g)",
          "vs kaitofukami\n(gate only, 16g)", "vs kaitofukami\n(combined, 32g)"]

def win_rate(margins):
    return 100 * sum(1 for m in margins if m > 0) / len(margins)

values = [win_rate(gate_boatlee), win_rate(gate_boatlee + fresh_boatlee),
          win_rate(gate_kaito), win_rate(gate_kaito + fresh_kaito)]
colors = [SLATE, TEAL, SLATE, TEAL]

fig, ax = plt.subplots(figsize=(8.5, 4.5))
bars = ax.bar(labels, values, color=colors, width=0.55)
for bar, v in zip(bars, values):
    ax.annotate(f"{v:.1f}%", (bar.get_x()+bar.get_width()/2, bar.get_height()),
                ha="center", va="bottom", fontsize=11, fontweight="bold")
ax.set_ylim(0, 105)
ax.set_ylabel("game-level win rate (%)")
ax.set_title("Win rate: original gate-only claim vs corrected combined figure", fontsize=13, fontweight="bold", color=INK, loc="left")
fig.tight_layout()
plt.show()
```

```python
# ---- Chart 2: per-game margins, all 32 games vs boatlee (gate + fresh) ----
gate_lbls  = [f"g{i+1}" for i in range(16)]
fresh_lbls = [f"f{i+1}" for i in range(16)]
all_lbls = gate_lbls + fresh_lbls
all_margins = gate_boatlee + fresh_boatlee
colors = [GREEN if m > 0 else RED for m in all_margins]

fig, ax = plt.subplots(figsize=(12, 4.5))
ax.bar(np.arange(len(all_margins)), all_margins, color=colors, width=0.7)
ax.axhline(0, color=SLATE, linewidth=1)
ax.axvline(15.5, color=INK, linewidth=1, linestyle="--", alpha=0.4)
ax.text(7.5, ax.get_ylim()[1]*0.85, "gate family", ha="center", fontsize=10, color=SLATE)
ax.text(23.5, ax.get_ylim()[1]*0.85, "fresh seeds", ha="center", fontsize=10, color=SLATE)
ax.set_xticks(np.arange(len(all_lbls)))
ax.set_xticklabels(all_lbls, fontsize=8)
ax.set_ylabel("margin (ShiV17 \u2212 boatlee)")
ax.set_title("All 32 games vs boatlee V16-RC2 \u2014 28W / 4L (87.5%), mean \u0394 +175", fontsize=13, fontweight="bold", color=INK, loc="left")
fig.tight_layout()
plt.show()
```

```python
# ---- Chart 3: per-game margins, all 32 games vs kaitofukami (gate + fresh) ----
all_margins_k = gate_kaito + fresh_kaito
colors_k = [GREEN if m > 0 else RED for m in all_margins_k]

fig, ax = plt.subplots(figsize=(12, 4.5))
ax.bar(np.arange(len(all_margins_k)), all_margins_k, color=colors_k, width=0.7)
ax.axhline(0, color=SLATE, linewidth=1)
ax.axvline(15.5, color=INK, linewidth=1, linestyle="--", alpha=0.4)
ax.text(7.5, ax.get_ylim()[1]*0.85, "gate family", ha="center", fontsize=10, color=SLATE)
ax.text(23.5, ax.get_ylim()[1]*0.85, "fresh seeds", ha="center", fontsize=10, color=SLATE)
ax.set_xticks(np.arange(len(all_lbls)))
ax.set_xticklabels(all_lbls, fontsize=8)
ax.set_ylabel("margin (ShiV17 \u2212 kaitofukami)")
ax.set_title("All 32 games vs kaitofukami v27 \u2014 29W / 3L (90.6%), mean \u0394 +3,348", fontsize=13, fontweight="bold", color=INK, loc="left")
fig.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ShiV17 color palette (keep it consistent)
INDIGO  = "#6d28d9"
VIOLET  = "#8b5cf6"
EMERALD = "#10b981"
CORAL   = "#f43f5e"
AMBER   = "#f59e0b"
SLATE   = "#64748b"
PAGE    = "#f8fafc"

plt.rcParams.update({
    "figure.facecolor": PAGE,
    "axes.facecolor": PAGE,
    "axes.edgecolor": "#cbd5e1",
    "axes.labelcolor": "#1e293b",
    "text.color": "#1e293b",
    "xtick.color": "#475569",
    "ytick.color": "#475569",
    "font.size": 12,
    "axes.grid": True,
    "grid.color": "#e2e8f0",
    "grid.linewidth": 0.8,
    "axes.spines.top": False,
    "axes.spines.right": False,
})
print("Ship it.")

# ---- Chart 1: win rate, before (V16a) vs after (ShiV17) ----
opponents = ["boatlee V16-RC2", "kaitofukami v27"]
v16a_wr = [0.0, 0.0]                    # 0W / 16L each
shiv17_wr = [93.8, 93.8]                # 15W / 1L each

x = np.arange(len(opponents))
w = 0.34
fig, ax = plt.subplots(figsize=(9, 5))
b1 = ax.bar(x - w/2, v16a_wr, w, label="V16a (before)", color=CORAL, edgecolor="#be123c", linewidth=1)
b2 = ax.bar(x + w/2, shiv17_wr, w, label="ShiV17 (after)", color=EMERALD, edgecolor="#047857", linewidth=1)

for bars in (b1, b2):
    for bar in bars:
        ax.annotate(f"{bar.get_height():.1f}%",
                    (bar.get_x() + bar.get_width()/2, bar.get_height()),
                    ha="center", va="bottom", fontsize=12, fontweight="bold")

ax.set_xticks(x)
ax.set_xticklabels(opponents)
ax.set_ylim(0, 110)
ax.set_ylabel("win rate (%)")
ax.set_title("Win rate vs the tape lords: 0% \u2192 93.8%", fontweight="bold", color=INDIGO, fontsize=15)
ax.legend(frameon=False, loc="upper left")
fig.tight_layout()
plt.show()


# ---- Chart 2: mean delta shift, before vs after ----
opponents = ["vs boatlee V16-RC2", "vs kaitofukami v27"]
v16a_d = [-14735, -13639]
shiv17_d = [176, 3587]

x = np.arange(len(opponents))
w = 0.34
fig, ax = plt.subplots(figsize=(9, 5.2))
b1 = ax.bar(x - w/2, v16a_d, w, label="V16a (before)", color=CORAL, edgecolor="#be123c", linewidth=1)
b2 = ax.bar(x + w/2, shiv17_d, w, label="ShiV17 (after)", color=EMERALD, edgecolor="#047857", linewidth=1)
ax.axhline(0, color="#475569", linewidth=1.1)

for bars in (b1, b2):
    for bar in bars:
        ax.annotate(f"{bar.get_height():+,.0f}",
                    (bar.get_x() + bar.get_width()/2, bar.get_height()),
                    ha="center", va="bottom", fontsize=11, fontweight="bold")

ax.set_xticks(x)
ax.set_xticklabels(opponents)
ax.set_ylabel("mean \u0394 (coins)")
ax.set_title("Mean score delta: from \u221214,735 to +176 / from \u221213,639 to +3,587",
             fontweight="bold", color=INDIGO, fontsize=14)
ax.legend(frameon=False, loc="upper left")
fig.tight_layout()
plt.show()

# The whole winning idea, in 10 lines (conceptual -- not the shipped agent)
RELAY_LEAD = 5                      # sell 5 turns before the scheduled sale, not 3

def should_counter_relay(step):
    # True when our dump lands 2 turns before boatlee's fixed relay
    future_step = step + RELAY_LEAD          # where our sale would land
    return (future_step % 24) == 17          # boatlee's scheduled-sell slot

# Which turns fire? (step 12 with lead 5 -> lands at 17; boatlee dumps at 14)
for step in range(10, 18):
    print(f"step {step:2d} (mod24=={step % 24:2d}): dump={should_counter_relay(step)}")
```
