## What I Learned From 2,800+ Games

| # | Lesson | The Hard Way |
|---|--------|-------------|
| 1 | **Verify your tar.gz** | One missing file = score 876 |
| 2 | **200 games minimum** | 20 games can't detect a 5% edge (p > 0.05) |
| 3 | **Win rate, not margin** | Elo only sees W/L/T. A $1 win = a $50K win |
| 4 | **Iterate, don't rebuild** | Each generation improves the last, never starts over |
| 5 | **Most games are ties** | Against similar agents, 80%+ games tie. Converting ties to wins is everything |
| 6 | **The last day matters most** | 85% of close losses happen in the final 7 steps |

## Build the Agent

No datasets, no internet, no GPU. The complete agent is embedded below.


```python
from pathlib import Path
import os, sys, time

WORKDIR = Path('/kaggle/working') if Path('/kaggle/working').is_dir() else Path.cwd()
```

```python
import base64, gzip, hashlib

EXPECTED_SHA256 = 'a5cb643b0516906fb7660613a4eebe00a6cc949a8de98d7744a7071038da824c'

# pipe-5: terminal planner boost + sell optimization
AGENT_GZ_B64 = ""

source_bytes = gzip.decompress(base64.b64decode(AGENT_GZ_B64))
actual = hashlib.sha256(source_bytes).hexdigest()
assert actual == EXPECTED_SHA256, f"SHA256 mismatch: {actual}"
print(f"Agent integrity verified: {actual[:16]}...")
print(f"Agent size: {len(source_bytes):,} bytes")

agent_path = WORKDIR / 'main.py'
agent_path.write_bytes(source_bytes)
print(f'Wrote {agent_path}')
```

```python
import ast

# Verify the agent parses and has the expected entry point
tree = ast.parse(source_bytes.decode('utf-8', errors='replace'))
functions = [node.name for node in ast.walk(tree) if isinstance(node, ast.FunctionDef)]
assert 'agent' in functions, f'No agent() function found! Functions: {functions[:10]}'
print(f'Verified: agent() found among {len(functions)} functions')
```

## Build the Submission Archive


```python
import gzip as gzip_mod, io, tarfile
from IPython.display import FileLink, display

ARCHIVE = WORKDIR / 'submission.tar.gz'

with ARCHIVE.open('wb') as fh:
    with tarfile.open(fileobj=fh, mode='w:gz') as tar:
        info = tarfile.TarInfo(name='main.py')
        info.size = len(source_bytes)
        info.mtime = int(time.time())
        tar.addfile(info, io.BytesIO(source_bytes))

sz = ARCHIVE.stat().st_size
print(f'Created {ARCHIVE} ({sz:,} bytes / {sz/1024:.0f} KB)')

# Verify
with tarfile.open(ARCHIVE) as tar:
    names = tar.getnames()
    print(f'Contents: {names}')
    assert 'main.py' in names, 'main.py missing from archive!'

print('Ready to submit!')
display(FileLink(str(ARCHIVE)))
```
