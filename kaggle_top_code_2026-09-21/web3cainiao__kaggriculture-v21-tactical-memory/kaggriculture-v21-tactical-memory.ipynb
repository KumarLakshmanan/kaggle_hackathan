## 3. The copy hypothesis, stated carefully

The runtime never asks *who* copied whom. It asks whether the two public farms
remain in the same production basin. In the canonical-id replay snapshot, the
strict public gate (board distance <=2 for 24 turns, money gap <=5) had ordered
market-intent equality of **99.73%**
on the outer replay split; among non-empty market turns it was
**99.39%**.

That is evidence of behavioral convergence, not proof of source-code copying.
Team display names are rename-sensitive, so offline splitting uses Kaggle's
team and submission ids. Those ids are excluded from `main.py` and serve only
to avoid mixing submissions during research.

There is also a hard limit: two identical deterministic agents receiving the
same public rules cannot both gain a permanent asymmetric advantage. Our live
v21-vs-v21 check is therefore expected to average zero. The target is lower
exploitability against today's population, not an impossible perpetual win over
an exact clone.

## 5. From 476 episodes to a 30-route tactical memory

Each current scoring submission contributes at most one vote:

1. choose its fit-only complete-action medoid;
2. for every step and item, record whether that route sells;
3. estimate `P(SELL item at t)` across 30 medoids;
4. retain the median positive quantity as a conservative scale.

```text
p_t(item) = (1 / 30) * sum_route 1[route sells item at t]
q_t(item) = median positive sold quantity at t
```

This creates **490 active SELL-hazard
steps**. Balancing by submission prevents a prolific team from dominating the
memory. At runtime this is a compact, nonparametric policy memory—30,603 bytes
for the whole agent—not a 13 GB replay database.

The memory is population-level. It stores no author identity and never chooses
a branch by team name.


````python
import pandas as pd
import matplotlib.pyplot as plt

route_screen = pd.DataFrame([
    ["Dennis Gioche medoid", 30, 39, 0.700000],
    ["Knight of Favonius medoid", 29, 39, 0.650000],
    ["2/3 rounded medoid", 27, 39, 0.600000],
    ["StopPlanting medoid", 26, 39, 0.600000],
    ["old v20/sash route", 9, 39, 0.210526],
], columns=["fit-only route", "wins", "games", "worst-seat win rate"])
route_screen["win rate"] = route_screen.wins / route_screen.games
display(route_screen)

ax = route_screen.plot.barh(
    x="fit-only route", y="win rate", legend=False, figsize=(9, 4.2), color="#0f766e"
)
ax.axvline(0.5, color="#64748b", linestyle="--", linewidth=1)
ax.set(xlabel="Route-validation win rate", ylabel="", xlim=(0, 1),
       title="A fresh route matters more than a more elaborate stale gate")
ax.grid(axis="x", alpha=0.25)
plt.tight_layout()
````

## 6. Why a neural SELL model was not promoted

We did train the obvious model first. A direct SELL-uplift MLP reached ROC AUC
`0.8185` and average precision `0.5996` on isolated counterfactual events. It
added one replay-holdout win—but when executed repeatedly against the adaptive
v20 artifact, it lost **9-31**.

The error was structural. A model trained on one isolated SELL intervention was
called many times inside one game; its own previous decisions changed inventory,
cash and future action feasibility. Event-level accuracy did not imply
closed-loop policy value.

v21 therefore promotes a complete, bounded option only after full-game W/L
testing. This is still data-driven: the hazard and quantity distributions are
estimated from 30 current submissions. It simply refuses to use a neural model
where the deployment distribution invalidates the label.


````python
option_screen = pd.DataFrame([
    ["safe current route", 38, 41, 0],
    ["v20-style SELL reorder", 33, 41, 0],
    ["full predicted quantity", 15, 41, 108],
    ["open full preemption", 14, 41, 564],
    ["half quantity + gate", 41, 41, 336],
], columns=["complete option", "wins", "games", "preemption activations"])
option_screen["win rate"] = option_screen.wins / option_screen.games
display(option_screen)

ax = option_screen.plot.barh(
    x="complete option", y="win rate", legend=False, figsize=(9, 4.2),
    color=["#94a3b8", "#f59e0b", "#dc2626", "#991b1b", "#0f766e"]
)
ax.axvline(0.5, color="#64748b", linestyle="--", linewidth=1)
ax.set(xlabel="Reserved option-validation win rate", ylabel="", xlim=(0, 1),
       title="More front-running is not better")
ax.grid(axis="x", alpha=0.25)
plt.tight_layout()
````

## 7. The selected option: one turn early, half the forecast

The safe route won **38/41**. Reordering existing SELLs in the old mirror style
fell to **33/41**. Selling the full consensus quantity one turn early collapsed
to **15/41**; removing the farm-similarity gate fell to **14/41**.

The selected option won **41/41**:

```python
if 24 <= step < 680 and public_farm_distance <= 2:
    if step - last_preempt >= 8:
        for item, probability, median_quantity in hazard[step + 1]:
            if probability >= 0.55 and item not in scheduled_sells:
                quantity = min(
                    own_legal_projected_shed[item],
                    30,
                    round(0.5 * median_quantity),
                )
                prepend_sell_if_positive(item, quantity)
```

It changed 336 market moments and 742 units across 41 validation games. Half
preemption converted three safe-control losses to wins and no wins to losses.
The interpretation is economic: take enough inventory forward to move first,
but leave enough cash/inventory continuity for the validated production route.

## 11. What still fails

- The 150-case outer still has six losses: four against the current Seb family,
  one PN case and one senkin13 case. One production route is not a universal
  closed-loop board policy.
- Replay counterfactuals do not let the recorded opponent react. The separate
  live tests reduce, but do not eliminate, that limitation.
- The first post-freeze set has only six candidate-side cases.
- v21 is neutral against its exact clone. Publishing the route and hazard will
  eventually make their edge more symmetric.
- Full preemption was catastrophic. A future model must predict complete
  short-horizon policy value, not merely whether an isolated SELL looks good.
- Tail risk remains: one live v20 game produced a large negative outlier even
  though the paired record was 39/40.

The next defensible step is a small portfolio of complete field+market options
selected by expected match value, trained with walk-forward groups and tested
against future executable submissions. The outer results in this Notebook must
not become that next model's tuning set.
